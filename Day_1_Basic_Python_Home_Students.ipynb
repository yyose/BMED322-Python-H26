## Python indexing

Indexing refers to the ability to access the specific elements of a sequence; for example, the different characters in a string.
Python indexes a string by assigning a different integer position to each of its characters. In other words, it identifies each character of a string with a different number. For example:

* In our natural language, the letter ``p`` in the word ``python`` would be assigned the **index 1**.
* However, in certain formal languages, such as Python, the first element of a string is indexed as **0, not 1**. Therefore, the character ``p`` in the string ``"python"`` has **index 0** and ``y`` has **index 1**. This is easier to understand visually, as shown in Figure 1 below.


## Python slicing

Slicing consists of extracting a sequence of characters (or **"slice"**) from a string. To slice a string, we use the operator ``[n:m]`` (*n-to-m*). This operator extracts a sequence of characters located from the “n-th” to the “m-th” index. The character assigned to the first index (n) will be included, but the character assigned to the second index (m) will be **excluded**.

For example:
```python
Input 1: my_string = "python"
Input 2: my_string[1:4]
Output: "yth"
```

### Adding and removing items from a list

There are three main functions we can use to **include** new elements in a list:
- The function ``list_name.append()`` adds **one item** to the end of a list.
> The argument of the function must be the item we want to add to the list. For example:
```python
number_list = [1, 4, 7]
number_list.append(8)
print(number_list)

Output: [1, 4, 7, 8]
```
- The function ``list_name.extend()`` adds **multiple items** to the end of a list.
> The argument of the function must be a list of items. For example:
```python
number_list = [1, 4, 7, 8]
number_list.extend([23, 5])
print(number_list)

Output: [1, 4, 7, 8, 23, 5]
```
- The function ``list_name.insert()`` inserts one item at a **desired position** of a list, as specified by its index.
> In this case, the argument is slightly more complex; first, we indicate the index position where we want to add our element; then, separated by a comma, we input the value of the item to insert. For example:
```python
number_list = [1, 4, 7, 8, 23, 5]
number_list.insert(3, 50) # We add number 50 in the position indexed as 3 (i.e., the 4th position)
print(number_list)

Output: [1, 4, 7, 50, 8, 23, 5]
```

We can use  ``while`` loops to repeat an indented block of code until a specific condition is met. ### Exercise 2: Transcribe a DNA sequence into mRNA

Given a DNA coding strand ``coding_DNA = "AAACCGTGCACCTACCAAATCTCCTTAGTGTAGGTTCTGACCGATTCGTG"``, write a program that transcribes the DNA sequence into an mRNA sequence. The mRNA sequence will have the same nucleotides as the DNA sequence, except for having all thymines ``"T"`` replaced with uracil ``"U"``. To do this, you must:
1. Create two variables: ``coding_DNA`` and ``mRNA``. Initially, assign an empty string to the variable ``mRNA`` (to do this, write: ``mRNA = ""``).
2. Inside a "for" loop that will read the individual characters of the string ``coding_DNA``, write a conditional statement that adds nucleotides into the empty mRNA string with every iteration of the loop; the conditional statement should include the following conditions:
    1. If the nucleotide in ``coding_DNA`` is either "A", "C" or "G", add the same nucleotide to the ``mRNA`` string (i.e., ``mRNA = mRNA + nucleotide``; in this case, "nucleotide" is the loop control variable).
    2. If the nucleotide in ``coding_DNA`` is "T", add the nucleotide "U" to the ``mRNA`` string (``mRNA = mRNA + "U"``; remember to use quotes around ``"U"`` because it is a string).

````python
# DNA til mRNA, der alle T => U
coding_DNA = "AAACCGTGCACCTACCAAATCTCCTTAGTGTAGGTTCTGACCGATTCGTG"
mRNA = ""
for c in coding_DNA:
    if c != "T":    # enten A, C eller G
        mRNA += c
    else:         # else
        mRNA += "U"
sortedmRNA = "".join(mRNA)
print("mRNA sequence:",sortedmRNA)  # do outside the loop!
print("DNA sequence:", coding_DNA)

# Fasit: print hver nukleotid
````

### Exercise 4: Estimate the GC-content of a DNA sequence

In this exercise, we will calculate the **GC-content** of a DNA sequence. The GC-content represents the percentage of nucleotides in a DNA sequence that are **guanine (G) or cytosine (C)**. This parameter is an indirect measure of the **thermal stability** of a DNA helix; the higher the GC-content, the higher the temperature needed to denaturalise the DNA's double strand.

Write a program including a **for loop** that reads every nucleotide in the sequence ``"CGCGATTAACGCGCTAGCTAGCTAGCTAAGTAATA"`` and increases the value of a variable ``GC_count = 0`` by ``1`` if a G or C is detected. Then, calculate the GC-content of the strand by dividing ´´GC_count´´ by the length of the DNA strand, and print the GC-content of the strand.
> To calculate the length of the DNA strand, we can use the built-in ``len()`` function; to learn about this function and familiarise yourself with external resources, please check the [Official Python Documentation](https://docs.python.org/3/library/functions.html) and [W3Schools](https://www.w3schools.com/python/ref_func_len.asp).

> Try to transform the GC content into a percentage, which will be extremely useful for the final exercise of Day 1.

````python
#indicator of thermal stability, GC => PCR => higher stability => PRIMER DESIGN
seq = "CGCGATTAACGCGCTAGCTAGCTAGCTAAGTAATA"
GC_count = 0
for b in seq:
    if b == "C" or b == "G":
        GC_count += 1
GC_percentage = (GC_count / len(seq))*100                  # useful for final exercise... percentage.
print(f"GC content in sequence: {GC_percentage}%"); print(type(GC_percentage))
````

* The [n:m] operator helps us create slices of strings and lists; the *m-th* element will **not** be included in the slice.
* Lists, written in brackets, can store multiple items separated by commas. Lists can be manipulated using multiple functions.
* Combining conditional statements and "for" loops is a powerful way to optimize your code.
* "While" loops are relatively simple, but can easily lead to infinite loops.
